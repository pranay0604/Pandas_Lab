```python
import pandas as pd
import numpy as np

data1 = {
    'emp_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'full_name': ['Alice Smith', 'Bob Johnson', 'Charlie Williams', 'Diana Brown', 'Eve Jones', 
                  'Frank Garcia', 'Grace Miller', 'Henry Davis', 'Ivy Rodriguez', 'Jack Martinez'],
    'department': ['HR', 'Engineering', 'IT', 'HR', 'Engineering', 
                   'Sales', 'IT', 'Engineering', 'Sales', 'IT'],
    'hire_date': pd.to_datetime(['2020-01-15', '2018-03-10', '2021-06-25', '2019-11-01', '2022-04-30',
                                 '2020-08-22', '2021-09-15', '2018-05-20', '2022-01-05', '2021-03-12']),
    'salary': [60000, 80000, 75000, 62000, 81000, 
               70000, 78000, 82000, 71000, 76000]
}
df1 = pd.DataFrame(data1)
print("DataFrame 1 (df1 - Personal Info):")
print(df1)
print("-" * 50)


data2 = {
    'emp_id': [101, 102, 105, 107, 111, 112],
    'department': ['HR', 'Engineering', 'Engineering', 'IT', 'Sales', 'HR'],
    'contact_email': ['alice@comp.com', 'bob@comp.com', 'eve@comp.com', 'grace@comp.com', 
                      'new111@comp.com', 'new112@comp.com'],
    'project_code': ['P1', 'P2', 'P2', 'P3', 'P4', 'P1']
}
df2 = pd.DataFrame(data2)
print("DataFrame 2 (df2 - Contact Info):")
print(df2)
print("-" * 50)
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on='emp_id',how='inner')
```

```python
pd.merge(df1,df2,on='emp_id',how='left')
```

```python
pd.merge(df1,df2,on='emp_id',how='right')
```

```python
pd.merge(df1,df2,on='emp_id',how='outer')
```

```python
df1, df2
```

```python
df1.join(
    df2.set_index('emp_id'),
    on='emp_id',
    how='left',
    lsuffix='_df1',
    rsuffix='_df2'
)
```

```python
df1.merge(df2, on='emp_id', how='left')
```

```python
pd.concat([df1.set_index('emp_id'), df2.set_index('emp_id')], axis=1)
```
